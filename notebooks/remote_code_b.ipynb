Two vision-language families whose makers ship the model code in the model repository rather than in transformers: [Phi-4-reasoning-vision-15B](https://huggingface.co/microsoft/Phi-4-reasoning-vision-15B) and [Molmo2-O-7B](https://huggingface.co/allenai/Molmo2-O-7B). That code was written against transformers 4.56 and 4.57 (Phi-4's image processor subclasses a module transformers 5 rewrote), so this card installs transformers 4.57.6 and works through its own queue, `runs/queue_remote_code_b.json` (a second card beside `remote_code.ipynb`; one queue per card, so no run starts twice) in the project repo ([AbstractPhil/qwen-symbolic-thinking-image-training](https://huggingface.co/AbstractPhil/qwen-symbolic-thinking-image-training)): first the base screens (the bare, shown and specification forms over the same 60 pictures and 40 specifications as every other family), then the arms under Lucia v1's recipe. Each spec names the repository code (`"trust_remote_code": true`), the commit it runs (`"revision"`) and this transformers release (`"transformers": "==4.57.6"`), so a card on transformers 5 leaves these runs alone and this card leaves every other run alone. Each run leaves `runs/done/<name>.json`; the cell ends after 45 idle minutes.

Runtime: a GPU (the RTX PRO 6000 fits both in bf16) and the `HF_TOKEN` secret with write access to the project repo. Run C1, restart the session (Runtime > Restart session), then run C2.

```python
# C1: install, then Runtime > Restart session
import subprocess, sys

def pip(*args):
    subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", *args])

# The transformers release the two families' repository code was built on (Phi-4-reasoning-vision on 4.56, Molmo2 on 4.57).
pip("git+https://github.com/AbstractEyes/qwen-symbolic-thinking@v0.1.35", "transformers==4.57.6", "accelerate>=1.0", "peft>=0.18")

try:                # Colab's torchao 0.10 breaks PEFT when it builds a LoRA (PEFT wants 0.16 or newer, or none at all)
    import torchao
    _v = tuple(int(x) for x in torchao.__version__.split(".")[:2])
    if _v < (0, 16):
        subprocess.call([sys.executable, "-m", "pip", "uninstall", "-y", "-q", "torchao"])
        print("torchao", torchao.__version__, "removed: PEFT cannot build a LoRA beside it")
    else:
        print("torchao", torchao.__version__, "stays")
except ImportError:
    print("no torchao")

print("installed; now Runtime > Restart session, then run C2")
```
